```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Kiran want to know the famous foods in each state of India. So, he opened Google and search for this question. Google showed that
in Delhi it is Chaat, in Gujarat it is Dal Dhokli, in Tamilnadu it is Pongal, in Andhrapradesh it is Biryani, in Assam it is Papaya Khar,
in Bihar it is Litti Chowkha and so on for all other states
"""

doc = nlp(text)

# List for storing all the names
all_gpe_names = []

for ent in doc.ents:
    if ent.label_ =="GPE":
        all_gpe_names.append(ent.text)
print(all_gpe_names)
print("Total number of GPE names in the text is:", len(all_gpe_names))
```

```python
text = """ 
Sachin Tendulkar was born on 24 April 1973, Virat Kholi was born on 5 November 1988, Dhoni was born on 7 July 1981
and finally Ricky ponting was born on 19 December 1974.
"""

doc = nlp(text)

all_birth_dates = []

for ent in doc.ents:
    if ent.label_ =="DATE":
        all_birth_dates.append(ent.text)

print(all_birth_dates)
print("Total number of birth dates in the text is:", len(all_birth_dates))
```
