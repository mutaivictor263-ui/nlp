Sentence & Word Tokenization In Spacy

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")
```

```python
for sentence in doc.sents:
    print(sentence)
```

```python
for sentence in doc.sents:
    for word in sentence:
        print(word)
```

##### Sentence & Word Tokenization In NLTK

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize

text = "Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi"
sent_tokenize(text)
```

```python
from nltk.tokenize import word_tokenize

text = ("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")

word_tokenize(text)
```

##### From above code you can see that Spacy is object orientad whereas NLTK is a string processing library
